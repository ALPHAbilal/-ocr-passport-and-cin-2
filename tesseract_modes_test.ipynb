# Tesseract Modes Test — Cropped Card

Uses the already-cropped `front_cropped.jpg` (856x540).
No card detection, no HSV, no perspective warp — just raw OCR on the clean crop.

Tests all PSM modes to find the best raw text output.

```python
import pytesseract
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# Load the already-cropped card — no detection needed
img = cv2.imread('front_cropped.jpg')
print(f'Loaded: {img.shape[1]}w x {img.shape[0]}h')

plt.figure(figsize=(12, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f'Cropped Card — {img.shape[1]}x{img.shape[0]}')
plt.axis('off')
plt.show()
```

---
## Test 1 — All PSM modes on original color image

No preprocessing. Just the raw crop through every PSM mode.

```python
pil_img = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

psm_modes = {
    3:  'Fully automatic page segmentation (default)',
    4:  'Single column of text',
    6:  'Single uniform block of text',
    11: 'Sparse text — find as much as possible',
    12: 'Sparse text with OSD',
}

for psm, desc in psm_modes.items():
    print('=' * 60)
    print(f'PSM {psm} — {desc}')
    print('=' * 60)
    try:
        text = pytesseract.image_to_string(pil_img, lang='ara+fra+eng', config=f'--psm {psm}')
        print(text.strip())
    except Exception as e:
        print(f'ERROR: {e}')
    print(f'\n({len(text.strip())} chars)')
    print()
```

---
## Test 2 — All PSM modes on grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
pil_gray = Image.fromarray(gray)

for psm, desc in psm_modes.items():
    print('=' * 60)
    print(f'PSM {psm} — Grayscale — {desc}')
    print('=' * 60)
    try:
        text = pytesseract.image_to_string(pil_gray, lang='ara+fra+eng', config=f'--psm {psm}')
        print(text.strip())
    except Exception as e:
        print(f'ERROR: {e}')
    print(f'\n({len(text.strip())} chars)')
    print()
```

---
## Test 3 — All PSM modes on adaptive threshold

```python
adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                  cv2.THRESH_BINARY, 21, 10)
pil_adaptive = Image.fromarray(adaptive)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('Original')
axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Grayscale')
axes[2].imshow(adaptive, cmap='gray')
axes[2].set_title('Adaptive Threshold')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()

for psm, desc in psm_modes.items():
    print('=' * 60)
    print(f'PSM {psm} — Adaptive — {desc}')
    print('=' * 60)
    try:
        text = pytesseract.image_to_string(pil_adaptive, lang='ara+fra+eng', config=f'--psm {psm}')
        print(text.strip())
    except Exception as e:
        print(f'ERROR: {e}')
    print(f'\n({len(text.strip())} chars)')
    print()
```

---
## Test 4 — All PSM modes on Otsu threshold

```python
_, otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
pil_otsu = Image.fromarray(otsu)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].imshow(otsu, cmap='gray')
axes[0].set_title('Otsu')
axes[1].imshow(adaptive, cmap='gray')
axes[1].set_title('Adaptive')
for ax in axes:
    ax.axis('off')
plt.suptitle('Otsu vs Adaptive', fontweight='bold')
plt.tight_layout()
plt.show()

for psm, desc in psm_modes.items():
    print('=' * 60)
    print(f'PSM {psm} — Otsu — {desc}')
    print('=' * 60)
    try:
        text = pytesseract.image_to_string(pil_otsu, lang='ara+fra+eng', config=f'--psm {psm}')
        print(text.strip())
    except Exception as e:
        print(f'ERROR: {e}')
    print(f'\n({len(text.strip())} chars)')
    print()
```

---
## Test 5 — Language combinations (best PSM from above)

Change `BEST_PSM` below to whichever PSM gave the best results above.

```python
BEST_PSM = 11  # change this based on results above

lang_combos = [
    'ara+fra+eng',
    'ara+fra',
    'fra+eng',
    'ara+eng',
    'ara',
    'fra',
    'eng',
]

for lang in lang_combos:
    print('=' * 60)
    print(f'Lang: {lang}  |  PSM: {BEST_PSM}')
    print('=' * 60)
    try:
        text = pytesseract.image_to_string(pil_img, lang=lang, config=f'--psm {BEST_PSM}')
        print(text.strip())
    except Exception as e:
        print(f'ERROR: {e}')
    print(f'\n({len(text.strip())} chars)')
    print()
```
